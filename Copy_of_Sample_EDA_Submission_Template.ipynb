## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Data Wrangling — titles
# Remove duplicate rows
titles.drop_duplicates(inplace=True)

# Convert numerical columns to numeric datatype
titles['release_year'] = pd.to_numeric(titles['release_year'], errors='coerce')
titles['runtime'] = pd.to_numeric(titles['runtime'], errors='coerce')
titles['seasons'] = pd.to_numeric(titles['seasons'], errors='coerce')
titles['imdb_score'] = pd.to_numeric(titles['imdb_score'], errors='coerce')
titles['imdb_votes'] = pd.to_numeric(titles['imdb_votes'], errors='coerce')
titles['tmdb_popularity'] = pd.to_numeric(titles['tmdb_popularity'], errors='coerce')
titles['tmdb_score'] = pd.to_numeric(titles['tmdb_score'], errors='coerce')

# Remove extra spaces from text columns
text_columns = titles.select_dtypes(include='object').columns

for col in text_columns:
    titles[col] = titles[col].str.strip()

# Handle missing values
titles['age_certification'] = titles['age_certification'].fillna('Not Rated')

titles['imdb_score'] = titles['imdb_score'].fillna(
    titles['imdb_score'].median()
)

titles['tmdb_score'] = titles['tmdb_score'].fillna(
    titles['tmdb_score'].median()
)

titles['imdb_votes'] = titles['imdb_votes'].fillna(0)

titles['tmdb_popularity'] = titles['tmdb_popularity'].fillna(0)

titles['seasons'] = titles['seasons'].fillna(0)
titles['description'] = titles['description'].fillna('Not Available')
titles['imdb_id'] = titles['imdb_id'].fillna('Not Available')
````

````python
# Data Wrangling — credits
# Remove duplicate rows
credits.drop_duplicates(inplace=True)

# Remove extra spaces from text columns
text_columns = credits.select_dtypes(include='object').columns

for col in text_columns:
    credits[col] = credits[col].str.strip()

# Handle missing character names
credits['character'] = credits['character'].fillna('Not Available')

# Standardize role values
credits['role'] = credits['role'].str.upper()
````

````python
# Final cleaning check - Titles
print("===== TITLES DATASET =====")
print("Shape:", titles.shape)
print("Duplicate rows:", titles.duplicated().sum())
print("\nMissing values:")
print(titles.isnull().sum())

# Final cleaning check - Credits
print("\n===== CREDITS DATASET =====")
print("Shape:", credits.shape)
print("Duplicate rows:", credits.duplicated().sum())
print("\nMissing values:")
print(credits.isnull().sum())
````

````python
merged_df = pd.merge(
    titles,
    credits,
    on='id',
    how='left'
)
````
